```python
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("AAPL.csv")
df.head()
```

```python
df.dtypes
```

```python
df["Date"]=pd.to_datetime(df["Date"])
```

```python
df.dtypes
```

```python
df.isna().sum()
```

```python
df1=df.copy()

X=df1.drop(columns=["Date","close","ingested_at_utc"],axis=1)
Y=df["close"]
```

```python
X.head()
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()

x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)

```

```python
x_train_scaled
```

```python
x_test_scaled
```

```python
import torch
import torch.nn as nn
```

```python
x_train_tensor=torch.tensor(x_train_scaled,dtype=torch.float32)
y_train_tensor=torch.tensor(y_train.values,dtype=torch.float32).view(-1,1)


x_test_tensor=torch.tensor(x_test_scaled,dtype=torch.float32)
y_test_tensor=torch.tensor(y_test.values,dtype=torch.float32).view(-1,1)
```

```python
x_train_tensor
```

```python
y_train_tensor
```

```python
from torch.utils.data import TensorDataset,DataLoader 
```

```python
train_data=TensorDataset(x_train_tensor,y_train_tensor)
test_data=TensorDataset(x_test_tensor,y_test_tensor)
```

```python
train_loader=DataLoader(train_data,batch_size=32,shuffle=True)
test_loader=DataLoader(test_data,batch_size=32)
```

# Ann

```python
class ANN(nn.Module):
    def __init__(self):
        super(ANN,self).__init__()

        self.model=nn.Sequential(
            nn.Linear(x_train.shape[1],32),
            nn.ReLU(),

            nn.Linear(32,32),
            nn.ReLU(),

            #Output
            nn.Linear(32,1)

        )

    def forward(self,x):
        return self.model(x)
        
```

```python
import torch.optim as optim
model=ANN()

criterion=nn.MSELoss()
optimizer=optim.Adam(model.parameters())
```

```python
epochs=2000
epoch=0

train_loss=[]
val_loss=[]
valid_score=float("inf")

for epoch in range(epochs):
    model.train()
    

    train_epoch_loss=0.0

    for xb,yb in train_loader:
        optimizer.zero_grad()
        
        output=model(xb)
        loss=criterion(output,yb)
        loss.backward()
        optimizer.step()

        train_epoch_loss+=loss.item()

    avg_train_loss=train_epoch_loss/len(train_loader)
    train_loss.append(avg_train_loss)


    #Validation
    model.eval()
    val_epoch_loss=0.0

    with torch.no_grad():
        for xb,yb in train_loader:
            output=model(xb)
            loss=criterion(output,yb)

        val_epoch_loss+=loss.item()

    avg_val_loss=val_epoch_loss/len(train_loader)
    val_loss.append(avg_val_loss)

    print(f"Epoch {epoch+1}/epochs trained ==> Train Loss = {avg_train_loss} ==> Validation Loss = {val_epoch_loss}")

    if avg_val_loss<valid_score:
        valid_score=avg_val_loss
        torch.save(model.state_dict(),"best_model.pt")
        
```

```python
model.load_state_dict(torch.load("best_model.pt"))
```

```python
model.eval()

with torch.no_grad():
    train_preds=model(x_train_tensor)
    test_preds=model(x_test_tensor)

    train_mse_loss=criterion(train_preds,y_train_tensor)
    test_mse_loss=criterion(test_preds,y_test_tensor)


print("training loss ",train_mse_loss.item())
print("testing mse loss ",test_mse_loss.item())


```
